```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data_radiation = pd.read_csv('downloads/radiation.csv')
```

```python
radiation_data = data_radiation.loc[1:,"timestamp":"cps_err"]
```

```python
#Radiation Frequency levels
y1 = radiation_data['cps'] 
plt.hist(y1,bins = 4, label = 'CPS')
plt.ylabel('Frequency')
plt.xlabel('CPS')
plt.title('CPS Frequency')
plt.show()
```

```python
y1 = radiation_data['cps'] 
x1 = radiation_data['timestamp'] 
y1err= radiation_data['cps_err']
# plot:
plt.errorbar(x1, y1, y1err, fmt='o', linewidth=1, capsize=4)
plt.plot(x1, y1, label = 'CPS', color ='tab:blue')
plt.xlabel('Time')
plt.ylabel('Counts per 10 seconds')
plt.ylim(-.05,0.75)
plt.title('Counts per 10 seconds vs Time')
plt.figure(figsize=(10, 7))
plt.show()
```

```python
print(f"The mean for our Radiation data is {np.mean(radiation_data['cps'])}")
print(f"The standard deviation for our Radiation data is {np.std(radiation_data['cps'])}")
print(f"The theoretical uncertainty for our Radiation data is {np.sqrt(np.mean(radiation_data['cps']))}")
```
