```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data_no_shield = pd.read_csv('downloads/no_shield.csv')
data_one_al_shield = pd.read_csv('downloads/one_al_shield.csv')
data_two_al_shield = pd.read_csv('downloads/two_al_shield.csv')
data_three_al_shield = pd.read_csv('downloads/three_al_shield.csv')
```

```python
no_shield_data = data_no_shield.loc[2:,"Time":"Counts"]
one_al_shield_data = data_one_al_shield.loc[2:,"Time":"Counts"]
two_al_shield_data = data_two_al_shield.loc[2:,"Time":"Counts"]
three_al_shield_data = data_three_al_shield.loc[2:,"Time":"Counts"]
```

```python
y1 = no_shield_data['Counts'] 
y2 = one_al_shield_data['Counts'] 
y3 = two_al_shield_data['Counts'] 
y4 = three_al_shield_data['Counts'] 
plt.hist(y1, label = 'No Shield')
plt.hist(y2, label = 'One Al Shield')
plt.hist(y3, label = 'Two Al Shield')
plt.hist(y4, label = 'Three Al Shield')
plt.xlabel('Counts per 10 seconds')
plt.ylabel('Frequency')
plt.title('Aluminium Shield- Counts Frequency')
plt.legend()
plt.show()
```

```python
print(f"The mean for our No Shield data is {np.mean(no_shield_data['Counts'])}")
print(f"The standard deviation for our No Shield data is {np.std(no_shield_data['Counts'])}")
print(f"The theoretical uncertainty for our No Shield data is {np.sqrt(np.mean(no_shield_data['Counts']))}")

print(f"The mean for our One Al Shield data is {np.mean(one_al_shield_data['Counts'] )}")
print(f"The standard deviation for our One Al Shield data is {np.std(one_al_shield_data['Counts'] )}")
print(f"The theoretical uncertainty for our One Al Shield data is {np.sqrt(np.mean(one_al_shield_data['Counts'] ))}")

print(f"The mean for our Two Al Shield data is {np.mean(two_al_shield_data['Counts'])}")
print(f"The standard deviation for our Two Al Shield data is {np.std(two_al_shield_data['Counts'])}")
print(f"The theoretical uncertainty for our Two Al Shield data is {np.sqrt(np.mean(two_al_shield_data['Counts']))}")

print(f"The mean for our Three Al Shield data is {np.mean(three_al_shield_data['Counts'])}")
print(f"The standard deviation for our Three Al Shield data is {np.std(three_al_shield_data['Counts'])}")
print(f"The theoretical uncertainty for our Three Al Shield data is {np.sqrt(np.mean(three_al_shield_data['Counts']))}")
```

```python
#How do the measured standard deviations compare to the theoretical uncertainty you
#would expect from counting statistics?
'''With the theoretical uncertainty being measured by taking the square root of the mean, 
the theoretical uncertainties compared to the standard deviations of the different measurements 
above are in good agreement with one another, as can be seen in Image 1, as the theoretical uncertainty
and the standard deviation for each data set using the aluminium for shielding yielded very similar results.'''
```

```python
#What general trend do you see in the counts measured as a function of the thickness of
#absorber material?
'''As can be seen in Histogram 1 above, the counts are the greatest when there is no 
aluminium shielding between the detector and the source.  Then, after one plate was added for 
shielding, the counts recorded dropped by roughly 100 counts over this interval, then when another 
plate was added, the counts continued to drop by another 50 counts, then after the third (last) 
sheet was added, the counts dropped by another 25 counts.  In other words, as the thickness of the 
shielding(absorber material) was increased, the counts recorded by the detector decreased. '''
```

```python
#Graphing different thickness
y1 = no_shield_data['Counts'] 
x1 = no_shield_data['Time'] 
y2 = one_al_shield_data['Counts']
x2 = one_al_shield_data['Time'] 
y3 = two_al_shield_data['Counts'] 
x3 = two_al_shield_data['Time'] 
y4 = three_al_shield_data['Counts'] 
x4 = three_al_shield_data['Time'] 
y1err= [np.sqrt(np.mean(no_shield_data['Counts']))]
y2err= [np.sqrt(np.mean(one_al_shield_data['Counts'] ))]
y3err= [np.sqrt(np.mean(two_al_shield_data['Counts']))]
y4err= [np.sqrt(np.mean(three_al_shield_data['Counts']))]
# plot:
plt.errorbar(x1, y1, y1err, fmt='o', linewidth=2, capsize=4)
plt.errorbar(x2, y2, y2err, fmt='o', linewidth=2, capsize=4)
plt.errorbar(x3, y3, y3err, fmt='o', linewidth=2, capsize=4)
plt.errorbar(x4, y4, y4err, fmt='o', linewidth=2, capsize=4)
plt.plot(x1, y1, label = 'No Shield', color ='tab:blue')
plt.plot(x2, y2, label = 'One Al Shield', color = 'orange')
plt.plot(x3, y3, label = 'Two Al Shield', color = 'green')
plt.plot(x4, y4, label = 'Three Al Shield',color = 'red')
plt.xlabel('Time')
plt.yscale('log')
plt.ylabel('Log 10 Scale counts per 10 seconds')
plt.title('Counts per 10 seconds vs Time')
plt.legend()
plt.show()
```

```python
# Do these graphs follow the trend you expect?
'''Yes, these results are consistent with the trend that we expected, 
since as the shielding thickness is increased, we know that the counts
recorded should decrease as the radiation interacts with the shielding
material, thus fewer counts are measured as the thickness increases.
Using the formula:
Absorption_coefficient = (ln(I1) - ln(I2)) / (x2 - x1)
Where we use the first and last data reading for aluminium, where at x1 = 0 and I1 = 366.91, and x2= 3 and I2 = 224.0 
'''
```

```python
#Do this same analysis for the second source. Do you expect the measured absorption coefficients
#to be the same? Why might you find differences?
'''
We do not expect the measured absorption coefficients to be the same, because tungsten has a higher density and atomic mass than aluminium, and because higher density materials are better at effectively blocking ionizing radiation, tungsten will thus be more effective at sheilding against the radiation.  Because tungsten will be better at blocking the radiation for these reasons, the absorption coefficient for tungsten should be higher than that of aluminium.

** Repeating the same analysis questions from above for the second source, each question has been copied and pasted below for organization and simplicity **
Plot histograms of the counts from each measurement for your second source all on the same graph. How do the mean (μ) and standard deviation (σ) compare?
The mean counts recorded for each measurement gets smaller after each new tungsten shield is added, increasing the radiation shielding thickness.  Likewise, the standard deviations also get smaller as more tungsten plates are added.  Thus, as the mean gets smaller because of the greater shielding thickness, so does the standard deviation.

How do the measured standard deviations compare to the theoretical uncertainty you would expect from counting statistics?
The measured standard deviations agree with the theoretical uncertainty, given that the theoretical uncertainty represents the square root of the mean, and as the mean gets smaller (due to increasing shielding thickness), the standard deviation and theoretical uncertainty get closer together and are very comparable throughout the experiment, as shown in Image 3.

What general trend do you see in the counts measured as a function of the thickness of absorber material? 
As shown in Image 3 and Histogram 2, as the thickness of the tungsten shield increases, the counts measured decreases.

Do these graphs follow the trend you expect? 
Yes, these graphs do follow the trend we expected, as we expected that as the tungsten shielding thickness was increased, the counts measured would decrease, and Image 4 supports this prediction.
Using the formula:
Absorption_coefficient_tungsten = (ln(I1) - ln(I2)) / (x2 - x1)
Where we use the first and last data reading for tungsten, where at x1 = 0 and I1 = 219.09, and x2= 3 and I2 = 77.36 

Absorption_coefficient_tungsten = (ln(219.09) - ln(77.36)) / (3 - 0) = 0.347
Thus, the absorption coefficient for tungsten is roughly 0.347 per tungsten plate.
'''
```

```python
no_w_shield = pd.read_csv('downloads/no_shield_unknown.csv')
one_w_shield = pd.read_csv('downloads/one_w_shield.csv')
two_w_shield = pd.read_csv('downloads/two_w_shield.csv')
three_w_shield = pd.read_csv('downloads/three_w_shield.csv')
```

```python
no_w_shield = no_w_shield.loc[2:,"Time":"Counts"]
one_w_shield = one_w_shield.loc[2:,"Time":"Counts"]
two_w_shield = two_w_shield.loc[2:,"Time":"Counts"]
three_w_shield = three_w_shield.loc[2:,"Time":"Counts"]
```

```python
y1 = no_w_shield['Counts'] 
y2 = one_w_shield['Counts'] 
y3 = two_w_shield['Counts'] 
y4 = three_w_shield['Counts'] 
plt.hist(y1, label = 'No Shield')
plt.hist(y2, label = 'One W Shield')
plt.hist(y3, label = 'Two W Shield')
plt.hist(y4, label = 'Three W Shield')
plt.xlabel('Counts')
plt.ylabel('Frequency')
plt.title('Counts Frequency')
plt.legend()
plt.show()
```

```python
no_w_shield_mean = np.mean(no_w_shield['Counts'])
no_w_shield_uncertainty = np.sqrt(no_w_shield_mean)
no_w_shield_std = np.std(no_w_shield['Counts'])


print(f"The theoretical uncertainty for the data lacking the tungsten shield (which has been derived from the mean of the measured counts) is {no_w_shield_uncertainty}.")
print(f"The standard deviation of the data lacking the tungsten shield is {no_w_shield_std}.")
print(f"The mean of the data lacking the tungsten shield is {no_w_shield_mean}.")

one_w_shield_mean = np.mean(one_w_shield['Counts'])
one_w_shield_uncertainty = np.sqrt(one_w_shield_mean)
one_w_shield_std = np.std(one_w_shield['Counts'])


print(f"The theoretical uncertainty for the data with one tungsten shield (which has been derived from the mean of the measured counts) is {one_w_shield_uncertainty}.")
print(f"The standard deviation of the data with one tungsten shield is {one_w_shield_std}.")
print(f"The mean of the data with one tungsten shield is {one_w_shield_mean}.")

two_w_shield_mean = np.mean(two_w_shield['Counts'])
two_w_shield_uncertainty = np.sqrt(two_w_shield_mean)
two_w_shield_std = np.std(two_w_shield['Counts'])


print(f"The theoretical uncertainty for the data with two tungsten shields (which has been derived from the mean of the measured counts) is {two_w_shield_uncertainty}.")
print(f"The standard deviation of the data set with two tungsten shields is {two_w_shield_std}.")
print(f"The mean of the data set with two tungsten shields is {two_w_shield_mean}.")

three_w_shield_mean = np.mean(three_w_shield['Counts'])
three_w_shield_uncertainty = np.sqrt(three_w_shield_mean)
three_w_shield_std = np.std(three_w_shield['Counts'])


print(f"The theoretical uncertainty for the data with three tungsten shields (which has been derived from the mean of the measured counts) is {three_w_shield_uncertainty}.")
print(f"The standard deviation of the data with three tungsten shields is {three_w_shield_std}.")
print(f"The mean of the data set with three tungsten shields is {three_w_shield_mean}.")
```
