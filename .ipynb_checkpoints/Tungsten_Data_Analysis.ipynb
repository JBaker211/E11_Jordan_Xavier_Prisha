```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
no_w_shield = pd.read_csv('no_shield_unknown.csv')
one_w_shield = pd.read_csv('one_w_shield.csv')
two_w_shield = pd.read_csv('two_w_shield.csv')
three_w_shield = pd.read_csv('three_w_shield.csv')
```

```python
no_w_shield = no_w_shield.loc[2:,"Time":"Counts"]
one_w_shield = one_w_shield.loc[2:,"Time":"Counts"]
two_w_shield = two_w_shield.loc[2:,"Time":"Counts"]
three_w_shield = three_w_shield.loc[2:,"Time":"Counts"]
```

```python
y1 = no_w_shield['Counts'] 
y2 = one_w_shield['Counts'] 
y3 = two_w_shield['Counts'] 
y4 = three_w_shield['Counts'] 
plt.hist(y1, label = 'No Shield')
plt.hist(y2, label = 'One W Shield')
plt.hist(y3, label = 'Two W Shield')
plt.hist(y4, label = 'Three W Shield')
plt.xlabel('Counts')
plt.ylabel('Frequency')
plt.title('Counts Frequency')
plt.legend()
plt.show()
```

```python

no_w_shield_mean = np.mean(no_w_shield['Counts'])
no_w_shield_uncertainty = np.sqrt(no_w_shield_mean)
no_w_shield_std = np.std(no_w_shield['Counts'])


print(f"The theoretical uncertainty for the data lacking the tungsten shield (which has been derived from the mean of the measured counts) is {no_w_shield_uncertainty}.")
print(f"The standard deviation of the data lacking the tungsten shield is {no_w_shield_std}.")
print(f"The mean of the data lacking the tungsten shield is {no_w_shield_mean}.")

one_w_shield_mean = np.mean(one_w_shield['Counts'])
one_w_shield_uncertainty = np.sqrt(one_w_shield_mean)
one_w_shield_std = np.std(one_w_shield['Counts'])


print(f"The theoretical uncertainty for the data with one tungsten shield (which has been derived from the mean of the measured counts) is {one_w_shield_uncertainty}.")
print(f"The standard deviation of the data with one tungsten shield is {one_w_shield_std}.")
print(f"The mean of the data with one tungsten shield is {one_w_shield_mean}.")

two_w_shield_mean = np.mean(two_w_shield['Counts'])
two_w_shield_uncertainty = np.sqrt(two_w_shield_mean)
two_w_shield_std = np.std(two_w_shield['Counts'])


print(f"The theoretical uncertainty for the data with two tungsten shields (which has been derived from the mean of the measured counts) is {two_w_shield_uncertainty}.")
print(f"The standard deviation of the data set with two tungsten shields is {two_w_shield_std}.")
print(f"The mean of the data set with two tungsten shields is {two_w_shield_mean}.")

three_w_shield_mean = np.mean(three_w_shield['Counts'])
three_w_shield_uncertainty = np.sqrt(three_w_shield_mean)
three_w_shield_std = np.std(three_w_shield['Counts'])


print(f"The theoretical uncertainty for the data with three tungsten shields (which has been derived from the mean of the measured counts) is {three_w_shield_uncertainty}.")
print(f"The standard deviation of the data with three tungsten shields is {three_w_shield_std}.")
print(f"The mean of the data set with three tungsten shields is {three_w_shield_mean}.")
```

```python
#Graphing different thickness
#Graphing different thickness
y1 = no_w_shield['Counts'] 
x1 = no_w_shield['Time'] 
y2 = one_w_shield['Counts']
x2 = one_w_shield['Time'] 
y3 = two_w_shield['Counts'] 
x3 = two_w_shield['Time'] 
y4 = three_w_shield['Counts'] 
x4 = three_w_shield['Time'] 
y1err= [np.sqrt(np.mean(no_w_shield['Counts']))]
y2err= [np.sqrt(np.mean(one_w_shield['Counts'] ))]
y3err= [np.sqrt(np.mean(two_w_shield['Counts']))]
y4err= [np.sqrt(np.mean(three_w_shield['Counts']))]
# plot:
plt.errorbar(x1, y1, y1err, fmt='o', linewidth=2, capsize=4)
plt.errorbar(x2, y2, y2err, fmt='o', linewidth=2, capsize=4)
plt.errorbar(x3, y3, y3err, fmt='o', linewidth=2, capsize=4)
plt.errorbar(x4, y4, y4err, fmt='o', linewidth=2, capsize=4)
plt.plot(x1, y1, label = 'No Shield', color = 'tab:blue')
plt.plot(x2, y2, label = 'One W Shield', color = 'orange')
plt.plot(x3, y3, label = 'Two W Shield', color = 'green')
plt.plot(x4, y4, label = 'Three W Shield', color = 'red')
plt.xlabel('Time')
plt.yscale('log')
plt.ylabel('Log 10 Scale counts per 10 seconds')
plt.title('Counts per 10 seconds vs Time')
plt.legend()
plt.show()
```
