```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
"""
PART 1
"""

# inside data analysis

inside_data = pd.read_csv('insidespecial - inside.csv')
inside_column_name_list = list(inside_data.columns)
inside_standard_deviations = []
inside_standard_means = []
inside_trial_uncs = []

for i in range(inside_data.shape[1]):
    inside_column_data = inside_data[inside_column_name_list[i]] 
    inside_standard_deviations.append(np.std(inside_column_data))
    inside_standard_means.append(inside_standard_deviations[i] / np.sqrt(len(inside_column_data)))

    # mean uncertainty for each inside data column
    inside_trial_unc =  np.sqrt(sum(inside_column_data))/(len(inside_column_data))
    inside_trial_uncs.append(inside_trial_unc)
    print(f"The mean uncertainty for {inside_column_name_list[i]} (for our inside data) is " + str(inside_trial_unc))

# outside data analysis

outside_data = pd.read_csv('outside2special - outside2.csv')
outside_column_name_list = list(outside_data.columns)
outside_standard_deviations = []
outside_standard_means = []
outside_trial_uncs = []

for i in range(outside_data.shape[1]):
    outside_column_data = outside_data[outside_column_name_list[i]]
    outside_standard_deviations.append(np.std(outside_column_data))
    outside_standard_means.append(outside_standard_deviations[i] / np.sqrt(len(outside_column_data)))

    # mean uncertainty for outside
    outside_trial_unc = np.sqrt(sum(outside_column_data))/(len(outside_column_data))
    outside_trial_uncs.append(outside_trial_unc)
    print(f"The mean uncertainty for {outside_column_name_list[i]} (for our outside data) is " + str(outside_trial_unc))


# average standard deviations
average_standard_deviation = np.sqrt(np.array(inside_standard_deviations)**2 + np.array(outside_standard_deviations[0:11])**2)

    
# data comparison 
for i in range(len(outside_standard_means)-1):
    if ((inside_standard_means[i] <= (3 * average_standard_deviation[i]+ inside_standard_means[i]) and (inside_standard_means[i] >= (inside_standard_means[i] - 3 * average_standard_deviation[i])))):
        inside_deviation = "Yes"
    else:
        inside_deviation = "No"
    if ((outside_standard_means[i] <= (3 * average_standard_deviation[i]+ outside_standard_means[i]) and (outside_standard_means[i] >= (outside_standard_means[i] - 3 * average_standard_deviation[i])))):
        outside_deviation = "Yes"
    else:
        outside_deviation = "No"
    if inside_deviation == "Yes" and outside_deviation == "Yes":
        print(f"{inside_column_name_list[i]}: Both inside and outside data are comparable.")
    elif inside_deviation == "Yes" and outside_deviation == "No":
        print(f"{inside_column_name_list[i]}: Inside data is significantly different from outside data.")
    elif inside_deviation == "No" and outside_deviation == "Yes":
        print(f"{inside_column_name_list[i]}: Outside data is significantly different from inside data.")   
    else:        
        print(f"{inside_column_name_list[i]}: Both inside and outside data are not comparable.")

for i in range(len(outside_standard_means)-1):
    # within or outside standard deviations
    # inside
    within_inside_calculation = inside_trial_uncs[i]
    print("sigma sub mu, inside: " + str(within_inside_calculation))
    
    # outside
    within_outside_calculation = outside_trial_uncs[i]
    print("sigma sub mu, outside: " + str(outside_trial_uncs[i]))
    
    # difference
    within_differences = abs(np.mean(inside_data[inside_column_name_list[i]]) - np.mean(outside_data[outside_column_name_list[i]]))
    print("difference: " + str(within_differences))
    
    # comparison 
    if within_inside_calculation > within_outside_calculation:
        comparison = within_differences/within_inside_calculation
    else:
        comparison = within_differences/within_outside_calculation

    # standard deviations away
    if comparison < 2.58:
        print(f"The {inside_column_name_list[i]} data for both inside and outside are within 1% of each other, indicating that the inside and outside data are comparable.") 
    else:
        print(f"The {inside_column_name_list[i]} data for both inside and outside are outside 1% of each other, indicating that the inside and outside data are not comparable.")
        print("As our increased amount of data makes differences in said data more pronounced, we believe that we have collected enough data to indicate whether it can be represented by a Gaussian distribution (which, it can't). This may be due to human error when changing locations to collect data, or due to insensitivity within our sensors.")


"""
PART 2
"""



# inside data analysis

inside_data = pd.read_csv('insidespecial - inside.csv')
inside_column_name_list = list(inside_data.columns)
inside_standard_deviations = []
inside_standard_means = []
inside_trial_uncs = []

for i in range(inside_data.shape[1]):
    inside_column_data = inside_data[inside_column_name_list[i]] 
    inside_standard_deviations.append(np.std(inside_column_data))
    inside_standard_means.append(inside_standard_deviations[i] / np.sqrt(len(inside_column_data)))

    # mean uncertainty for each inside data column
    inside_trial_unc = np.sqrt(sum(inside_column_data))/(len(inside_column_data))
    inside_trial_uncs.append(inside_trial_unc)
    print(f"The mean uncertainty for {inside_column_name_list[i]} (for our inside data) is " + str(inside_trial_unc))

# other group inside data analysis

other_inside_data = pd.read_csv('inside_5min - inside_5min.csv')
other_inside_column_name_list = list(other_inside_data.columns)
other_inside_standard_deviations = []
other_inside_standard_means = []
other_inside_trial_uncs = []

for i in range(other_inside_data.shape[1]):
    other_inside_column_data = other_inside_data[other_inside_column_name_list[i]]
    other_inside_standard_deviations.append(np.std(other_inside_column_data))
    other_inside_standard_means.append(other_inside_standard_deviations[i] / np.sqrt(len(other_inside_column_data)))

    # mean uncertainty for each inside data column
    other_inside_trial_unc = np.sqrt(sum(other_inside_column_data))/(len(other_inside_column_data))
    other_inside_trial_uncs.append(other_inside_trial_unc)
    print(f"The mean uncertainty for {inside_column_name_list[i]} (for the other group's inside data) is " + str(other_inside_trial_unc))

# average standard deviations
average_standard_deviation = np.sqrt(np.array(inside_standard_deviations)**2 + np.array(other_inside_standard_deviations)**2)

# data comparison 
for i in range(len(other_inside_standard_means)):
    if ((inside_standard_means[i] <= (3 * average_standard_deviation[i]+ inside_standard_means[i]) and (inside_standard_means[i] >= (inside_standard_means[i] - 3 * average_standard_deviation[i])))):
        inside_deviation = "Yes"
    else:
        inside_deviation = "No"
    if ((other_inside_standard_means[i] <= (3 * average_standard_deviation[i]+ other_inside_standard_means[i]) and (other_inside_standard_means[i] >= (other_inside_standard_means[i] - 3 * average_standard_deviation[i])))):
        other_inside_deviation = "Yes"
    else:
        other_inside_deviation = "No"
    if inside_deviation == "Yes" and other_inside_deviation == "Yes":
        print(f"{inside_column_name_list[i]}: Both inside and other group's inside data are comparable.")
    elif inside_deviation == "Yes" and other_inside_deviation == "No":
        print(f"{inside_column_name_list[i]}: Inside data is significantly different from other group's inside data.")
    elif inside_deviation == "No" and other_inside_deviation == "Yes":
        print(f"{inside_column_name_list[i]}: Other group's inside data is significantly different from inside data.")   
    else:        
        print(f"{inside_column_name_list[i]}: Both inside and other group's inside data are not comparable.")

for i in range(len(inside_standard_means)-1):
    # within or outside standard deviations
    # inside
    within_inside_calculation = inside_trial_uncs[i]
    print("sigma sub mu, inside: " + str(inside_trial_uncs[i]))
    
    # other inside
    within_other_inside_calculation = other_inside_trial_uncs[i]
    print("sigma sub mu, other group's inside: " + str(other_inside_trial_uncs[i]))
    
    # difference
    within_differences = abs(np.mean(inside_data[inside_column_name_list[i]]) - np.mean(other_inside_data[other_inside_column_name_list[i]]))
    print("difference: " + str(within_differences))
    
    # comparison 
    if within_inside_calculation > within_other_inside_calculation:
        comparison = within_differences/within_inside_calculation
    else:
        comparison = within_differences/within_other_inside_calculation

    # standard deviations away
    if comparison < 2.58:
        print(f"The {inside_column_name_list[i]} data for both inside and the other group's inside are within 1% of each other, indicating that the inside and other inside data are comparable.") 
    else:
        print(f"The {inside_column_name_list[i]} data for both inside and the other group's inside are outside 1% of each other, indicating that the inside and other inside data are not comparable.")
        print("Our sensor data is quite dissimilar, indicating that our sensors may ")


# outside data analysis

outside_data = pd.read_csv('outside2special - outside2.csv')
outside_column_name_list = list(outside_data.columns)
outside_standard_deviations = []
outside_standard_means = []
outside_trial_uncs = []

for i in range(outside_data.shape[1]):
    outside_column_data = outside_data[outside_column_name_list[i]] 
    outside_standard_deviations.append(np.std(outside_column_data))
    outside_standard_means.append(outside_standard_deviations[i] / np.sqrt(len(outside_column_data)))

    # mean uncertainty for outside
    outside_trial_unc = np.sqrt(sum(outside_column_data))/(len(outside_column_data))
    outside_trial_uncs.append(outside_trial_unc)
    print(f"The mean uncertainty for {outside_column_name_list[i]} (for our outside data) is " + str(outside_trial_unc))

# other group outside data analysis

other_outside_data = pd.read_csv('double_data_outside_5min - double_data_outside_5min.csv')
other_outside_column_name_list = list(other_outside_data.columns)
other_outside_standard_deviations = []
other_outside_standard_means = []
other_outside_trial_uncs = []

for i in range(other_outside_data.shape[1]):
    other_outside_column_data = other_outside_data[other_outside_column_name_list[i]]
    other_outside_standard_deviations.append(np.std(other_outside_column_data))
    other_outside_standard_means.append(other_outside_standard_deviations[i] / np.sqrt(len(other_outside_column_data)))

    # mean uncertainty for outside
    other_outside_trial_unc = np.sqrt(sum(other_outside_column_data))/(len(other_outside_column_data))
    other_outside_trial_uncs.append(other_outside_trial_unc)
    print(f"The mean uncertainty for {outside_column_name_list[i]} (for the other group's outside data) is " + str(other_outside_trial_unc))

# average standard deviations
average_standard_deviation = np.sqrt(np.array(outside_standard_deviations[0:11])**2 + np.array(other_outside_standard_deviations)**2)

# data comparison 
for i in range(len(other_outside_standard_means)):
    if ((outside_standard_means[i] <= (3 * average_standard_deviation[i]+ outside_standard_means[i]) and (outside_standard_means[i] >= (outside_standard_means[i] - 3 * average_standard_deviation[i])))):
        outside_deviation = "Yes"
    else:
        outside_deviation = "No"
    if ((other_outside_standard_means[i] <= (3 * average_standard_deviation[i]+ other_outside_standard_means[i]) and (other_outside_standard_means[i] >= (other_outside_standard_means[i] - 3 * average_standard_deviation[i])))):
        other_outside_deviation = "Yes"
    else:
        other_outside_deviation = "No"
    if outside_deviation == "Yes" and other_outside_deviation == "Yes":
        print(f"{outside_column_name_list[i]}: Both outside and other group's outside data are comparable.")
    elif outside_deviation == "Yes" and other_outside_deviation == "No":
        print(f"{outside_column_name_list[i]}: Outisde data is significantly different from other group's outside data.")
    elif outside_deviation == "No" and other_outside_deviation == "Yes":
        print(f"{outside_column_name_list[i]}: Other group's outside data is significantly different from outside data.")   
    else:        
        print(f"{outside_column_name_list[i]}: Both outside and other group's outside data are not comparable.")

for i in range(len(outside_standard_means)-1):
    # within or outside standard deviations
    # outside
    within_outside_calculation = outside_trial_uncs[i]
    print("sigma sub mu, outside: " + str(outside_trial_uncs[i]))
    
    # other outside
    within_other_outside_calculation = other_outside_trial_uncs[i]
    print("sigma sub mu, other group's outside: " + str(other_outside_trial_uncs[i]))
    
    # difference
    within_differences = abs(np.mean(outside_data[outside_column_name_list[i]]) - np.mean(other_outside_data[other_outside_column_name_list[i]]))
    print("difference: " + str(within_differences))
    
    # comparison 
    if within_outside_calculation > within_other_outside_calculation:
        comparison = within_differences/within_outside_calculation
    else:
        comparison = within_differences/within_other_outside_calculation
    
    # standard deviations away
    if comparison < 2.58:
        print(f"The {outside_column_name_list[i]} data for both outside and the other group's outside are within 1% of each other, indicating that the outside and the other outside data are comparable.") 
    else:
        print(f"The {outside_column_name_list[i]} data for both outside and the other group's outside are outside 1% of each other, indicating that the outside and the other outside data are not comparable.")
        print("As our increased amount of data makes differences in said data more pronounced, we believe that we have collected enough data to indicate whether it can be represented by a Gaussian distribution (which, it can't). This may be due to human error when changing locations to collect data, or due to insensitivity within our sensors. We don't believe our sensors had very similar data at all, which indicates that our sensors may be sensitive, but also generally inaccurate (due to how widely our data differed.")


"""
PART 3
"""

# outside data analysis

outside_data = pd.read_csv('outside2special - outside2.csv')
outside_column_name_list = list(outside_data.columns)
outside_standard_deviations = []
outside_standard_means = []
outside_trial_uncs = []

for i in range(outside_data.shape[1]):
    outside_column_data = outside_data[outside_column_name_list[i]] 
    outside_standard_deviations.append(np.std(outside_column_data))
    outside_standard_means.append(outside_standard_deviations[i] / np.sqrt(len(outside_column_data)))

    # mean uncertainty for outside
    outside_trial_unc = np.sqrt(sum(outside_column_data))/(len(outside_column_data))
    outside_trial_uncs.append(outside_trial_unc)
    print(f"The mean uncertainty for {outside_column_name_list[i]} (for our outside data) is " + str(outside_trial_unc))

# outside alley

other_outside_data = pd.read_csv('outside_alley.csv')
other_outside_column_name_list = list(other_outside_data.columns)
other_outside_standard_deviations = []
other_outside_standard_means = []
other_outside_trial_uncs = []

for i in range(other_outside_data.shape[1]):
    other_outside_column_data = other_outside_data[other_outside_column_name_list[i]]
    other_outside_standard_deviations.append(np.std(other_outside_column_data))
    other_outside_standard_means.append(other_outside_standard_deviations[i] / np.sqrt(len(other_outside_column_data)))

    # mean uncertainty for outside
    other_outside_trial_unc = np.sqrt(sum(other_outside_column_data))/(len(other_outside_column_data))
    other_outside_trial_uncs.append(other_outside_trial_unc)
    print(f"The mean uncertainty for {outside_column_name_list[i]} (for our outside data) is " + str(other_outside_trial_unc))

# average standard deviations
average_standard_deviation = np.sqrt(np.array(outside_standard_deviations[0:11])**2 + np.array(other_outside_standard_deviations)**2)

# data comparison 
for i in range(len(other_outside_standard_means)):
    if ((outside_standard_means[i] <= (3 * average_standard_deviation[i]+ outside_standard_means[i]) and (outside_standard_means[i] >= (outside_standard_means[i] - 3 * average_standard_deviation[i])))):
        outside_deviation = "Yes"
    else:
        outside_deviation = "No"
    if ((other_outside_standard_means[i] <= (3 * average_standard_deviation[i]+ other_outside_standard_means[i]) and (other_outside_standard_means[i] >= (other_outside_standard_means[i] - 3 * average_standard_deviation[i])))):
        other_outside_deviation = "Yes"
    else:
        other_outside_deviation = "No"
    if outside_deviation == "Yes" and other_outside_deviation == "Yes":
        print(f"{outside_column_name_list[i]}: Both outside and outside alley data are comparable.")
    elif outside_deviation == "Yes" and other_outside_deviation == "No":
        print(f"{outside_column_name_list[i]}: Outside data is significantly different from outside alley data.")
    elif outside_deviation == "No" and other_outside_deviation == "Yes":
        print(f"{outside_column_name_list[i]}: Outside alley data is significantly different from outside data.")   
    else:        
        print(f"{outside_column_name_list[i]}: Both data sets are not comparable.")

for i in range(len(outside_standard_means)-1):
    # within or outside standard deviations
    # outside
    within_outside_calculation = outside_trial_uncs[i]
    print("sigma sub mu, outside: " + str(outside_trial_uncs[i]))
    
    # other outside
    within_other_outside_calculation = other_outside_trial_uncs[i]
    print("sigma sub mu, other group's outside: " + str(other_outside_trial_uncs[i]))
    
    # difference
    within_differences = abs(np.mean(outside_data[outside_column_name_list[i]]) - np.mean(other_outside_data[other_outside_column_name_list[i]]))
    print("difference: " + str(within_differences))
    
    # comparison 
    if within_outside_calculation > within_other_outside_calculation:
        comparison = within_differences/within_outside_calculation
    else:
        comparison = within_differences/within_other_outside_calculation
    
    # standard deviations away
    if comparison < 2.58:
        print(f"The {outside_column_name_list[i]} data for both outside and outside alley are within 1% of each other, indicating that the outside and the other outside data are comparable.") 
    else:
        print(f"The {outside_column_name_list[i]} data for both outside and outside alley are outside 1% of each other, indicating that the outside and the other outside data are not comparable.")
        print("As our increased amount of data makes differences in said data more pronounced, we believe that we have collected enough data to indicate whether it can be represented by a Gaussian distribution (which, it can't). This may be due to human error when changing locations to collect data, or due to insensitivity within our sensors. We don't believe our sensors had very similar data at all, which indicates that our sensors may be sensitive, but also generally inaccurate (due to how widely our data differed.")
```
